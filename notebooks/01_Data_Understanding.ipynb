## Business Problem

Banks receive thousands of loan applications every day. Approving loans for high-risk customers can lead to financial losses, while rejecting eligible customers can reduce business opportunities.

The objective of this project is to build a machine learning model that predicts whether a customer is eligible for a loan based on banking history and CIBIL information. Additionally, the project recommends an appropriate loan amount, repayment tenure, and provides business insights using exploratory data analysis.


## 1. Import Libraries

```python
# Data Manipulation
import pandas as pd
import numpy as np

# Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 10)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully.")
```

## 2. Load Datasets

```python
import pandas as pd

internal_df = pd.read_excel("../data/raw/Internal_Bank_Dataset.xlsx")
external_df = pd.read_excel("../data/raw/External_Cibil_Dataset.xlsx")
unseen_df = pd.read_excel("../data/raw/Unseen_Dataset.xlsx")

print("Datasets loaded successfully.")
```

## 3. Inspect Dataset Shapes

```python
print("Internal Dataset Shape :", internal_df.shape)
print("External Dataset Shape :", external_df.shape)
print("Unseen Dataset Shape   :", unseen_df.shape)
```

## 4. Preview Data

```python
print("Internal Dataset")
display(internal_df.head())

print("External Dataset")
display(external_df.head())

print("Unseen Dataset")
display(unseen_df.head())
```

## 5. Data Types

```python
print("Internal Dataset")
display(internal_df.info())

print("External Dataset")
display(external_df.info())

print("Unseen Dataset")
display(unseen_df.info())
```

## 6. Missing Values

```python
missing_internal = internal_df.isnull().sum().sort_values(ascending=False)
missing_external = external_df.isnull().sum().sort_values(ascending=False)
missing_unseen = unseen_df.isnull().sum().sort_values(ascending=False)

print("Internal Dataset")
display(missing_internal[missing_internal > 0])

print("External Dataset")
display(missing_external[missing_external > 0])

print("Unseen Dataset")
display(missing_unseen[missing_unseen > 0])
```

## 7. Duplicate Records

```python
print("Internal Duplicate Rows :", internal_df.duplicated().sum())

print("External Duplicate Rows :", external_df.duplicated().sum())

print("Unseen Duplicate Rows :", unseen_df.duplicated().sum())
```

## 8. Merge Datasets

```python
merged_df = pd.merge(
    internal_df,
    external_df,
    on="PROSPECTID",
    how="inner"
)

print("Merged Dataset Shape :", merged_df.shape)

display(merged_df.head())
```

## 9. Statistical Summary

```python
display(internal_df.describe())

display(external_df.describe())

display(unseen_df.describe())
```

## 10. Save the Processed Dataset


```python
merged_df.to_csv("../data/processed/merged_dataset.csv", index=False)

print("Merged dataset saved successfully.")
```
